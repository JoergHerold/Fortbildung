## Abschluss – Besprechung einer Schülerlösung

Eine Lösung aus der Make-Phase auflegen und die vier Begriffe daran benennen lassen: Variable, Zuweisung, Eingabe/Ausgabe, Verkettung. Die Lösung schon beim Rundgang während Make auswählen und der Person Bescheid geben.

Besonders ergiebig sind:

- eine Rückmeldung, in der die eingelesene Variable nicht vorkommt (`print("Klingt gut!")` statt `print(fruehstueck + " klingt gut!")`) – daran wird sichtbar, wozu die Variable überhaupt dient;
- eine Lösung, die alles in ein einziges `input()` packt;
- bei Vorkenntnis-Schülern ein f-String (`f"Hallo {name}!"`) – als zweiten Weg benennen, aber erst in Lektion 2 erklären.

## Make – Beispiellösung

````python
name = input("Wie heißen Sie? ")
print("Hallo " + name + "!")
fruehstueck = input("Was hatten Sie zum Frühstück, " + name + "? ")
print(fruehstueck + " klingt gut!")
farbe = input("Was ist Ihre Lieblingsfarbe, " + name + "? ")
print(farbe + " ist auch meine Lieblingsfarbe!")
````

## Erweiterung – Beispiellösung

````python
name = input("Wie heißen Sie? ")
alter = int(input("Wie alt sind Sie, " + name + "? "))
neues_alter = alter + 1
print("Danke – dann werden Sie am nächsten Geburtstag " + str(neues_alter) + ".")
````

## Schlüsselbegriffe – ausgefüllt

| Begriff | Erklärung | Wie in Python umsetzen? |
|---|---|---|
| **Variable** | Behälter für einen Wert, der sich im Programm ändern kann | `name = "Alex"` |
| **Zuweisung** | Einer Variablen einen Wert geben | `zahl = 56` |
| **Eingabe / Ausgabe** | Eingabe über die Tastatur, Ausgabe auf dem Bildschirm | `antwort = input()`, `print(antwort)` |
| **Verkettung** | Mehrere Texte zu einem Text zusammenfügen | `print("Hallo " + name)` |
| **Leerzeile** | Eine Zeile ohne Inhalt ausgeben | `print()` |

**Wichtige Hinweise zu Python**

- Python unterscheidet Groß- und Kleinschreibung: `Name` und `name` sind verschieden.
- Ein `NameError` tritt auf, wenn ein Bezeichner nicht bekannt ist – häufig durch Tippfehler oder falsche Groß- und Kleinschreibung.
