## P/R – Starterprogramm

Das Startprogramm wie in der Schülerfassung:

```
# Starter-Programm Lesson 1
name = "Billy"
print("We want to know if you like progamming!")
print()
print("Do you like programming " + name + "?")
answer = input()
print("Great! You said " + answer + "!")
print("Let's learn some Python today")
```

## I – Musterantworten (Kurzfassung)

1. **Variablen:** `name` (String mit einem Namen), `answer` (speichert die Benutzereingabe).  
2. **Weitere Funktion:** `input()` – liest eine **Tastatureingabe** als **String** ein.  
3. **`+` Operator:** **String-Konkatenation** (Zusammenfügen von Text).  
4. **`print()` ohne Inhalt:** erzeugt eine **Leerzeile**.

## M – Lösungen zu den Änderungen 1–3
**1. Eigenen Namen einsetzen:**

```
# Starter-Programm Lesson 1
name = "IhrName"
print("We want to know if you like progamming!")
print()
print("Do you like programming " + name + "?")
answer = input()
print("Great! You said " + answer + "!")
print("Let's learn some Python today")
```

**2. Antwort in derselben Zeile eingeben:**

```
name = "Billy"
print("We want to know if you like progamming!")
print()
answer = input("Do you like programming " + name + "? ")
print("Great! You said " + answer + "!")
print("Let's learn some Python today")
```

**3. Zu Beginn nach dem Namen fragen:**

```
name = input("What is your name? ")
print("We want to know if you like progamming!")
print()
answer = input("Do you like programming " + name + "? ")
print("Great! You said " + answer + "!")
print("Let's learn some Python today")
```

## M – Beispielprogramm „Drei Fragen“

```
# Programm mit drei Fragen
name = input("What is your name? ")
print("Hello " + name)
print()
breakfast = input("What did you have for breakfast, " + name + "? ")
print("Great! " + breakfast + " sounds delicious.")
colour = input("What is your favourite colour, " + name + "? ")
print("My favourite colour is " + colour + " too!")
```

### Erweiterung – Beispiel
Berechnen des Alters beim nächsten Geburtstag:

```
# Erweiterung: Alter +1 ausgeben
name = input("What is your name? ")
age = int(input("How old are you " + name + "? "))
newage = age + 1
print("Thank you - that means you will be " + str(newage) + " next birthday")
```
