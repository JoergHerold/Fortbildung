## Teil 1: Sekunden umrechnen

Die Steuerung einer CNC-Fräse zählt die Laufzeit eines Auftrags in **Sekunden**. Für das Protokoll soll die Laufzeit in Stunden, Minuten und Sekunden angegeben werden.

Schreiben Sie ein Programm, das die Laufzeit in Sekunden einliest und umgerechnet ausgibt.

So könnte die Ausgabe aussehen:

```
Laufzeit in Sekunden: 7384
Das sind 2 h, 3 min und 4 s.
```

*Zum Nachdenken:* Wie viele Sekunden hat eine Stunde, wie viele eine Minute? Was bleibt übrig, nachdem Sie die ganzen Stunden herausgerechnet haben?

````python
laufzeit = int(input("Laufzeit in Sekunden: "))

stunden = laufzeit // 3600     # ganze Stunden (1 h = 3600 s)
rest = laufzeit % 3600         # Sekunden, die danach übrig bleiben
minuten = rest // 60           # ganze Minuten im Rest
sekunden = rest % 60           # übrige Sekunden

print(f"Das sind {stunden} h, {minuten} min und {sekunden} s.")
````

## Teil 2: Lange Aufträge

Manche Aufträge laufen über mehrere Tage. Erweitern Sie Ihr Programm so, dass zusätzlich die **Tage** ausgegeben werden.

So könnte die Ausgabe aussehen:

```
Laufzeit in Sekunden: 200000
Das sind 2 Tage, 7 h, 33 min und 20 s.
```

````python
laufzeit = int(input("Laufzeit in Sekunden: "))

tage = laufzeit // 86400       # 1 Tag = 24 * 3600 s = 86400 s
rest = laufzeit % 86400
stunden = rest // 3600
rest = rest % 3600
minuten = rest // 60
sekunden = rest % 60

print(f"Das sind {tage} Tage, {stunden} h, {minuten} min und {sekunden} s.")
````

## Teil 3: Wann ist die Maschine fertig?

Ein Auftrag wird zu einer bestimmten Uhrzeit gestartet und läuft eine bestimmte Anzahl von **Minuten**. Schreiben Sie ein Programm, das Startstunde, Startminute und Laufzeit einliest und die **Uhrzeit** ausgibt, zu der die Maschine fertig ist.

Achten Sie darauf, dass auch Aufträge über Mitternacht hinaus richtig berechnet werden.

So könnte die Ausgabe aussehen:

```
Startstunde: 22
Startminute: 45
Laufzeit in Minuten: 150
Die Maschine ist um 1:15 Uhr fertig.
```

Prüfen Sie Ihr Programm auch mit diesen Eingaben:

| Startstunde | Startminute | Laufzeit in Minuten | erwartete Stunde | erwartete Minute |
|---|---|---|---|---|
| 8 | 0 | 90 | 9 | 30 |
| 23 | 50 | 20 | 0 | 10 |
| 8 | 0 | 3000 | 10 | 0 |

````python
start_stunde = int(input("Startstunde: "))
start_minute = int(input("Startminute: "))
laufzeit = int(input("Laufzeit in Minuten: "))

# alles in Minuten seit Mitternacht des Starttags umrechnen
ende = start_stunde * 60 + start_minute + laufzeit

stunde = ende // 60 % 24       # % 24 sorgt für den Sprung über Mitternacht
minute = ende % 60

print(f"Die Maschine ist um {stunde}:{minute} Uhr fertig.")
````

*Hinweis zur Lösung:* Der Kern ist die Idee, Start und Laufzeit in **eine gemeinsame Einheit** (Minuten seit Mitternacht) zu bringen. `ende // 60` liefert dann die Stunden seit Mitternacht, z. B. 25 für 1 Uhr am Folgetag; `% 24` bringt das wieder in den Bereich 0 bis 23.

Beim dritten Testfall lautet die Ausgabe `10:0 Uhr`. Das ist rechnerisch richtig und genau der Anlass für Erweiterung 2.

### Erweiterung

2. Eine Digitaluhr zeigt `01:15` statt `1:15`. Im f-String erreichen Sie das mit `{stunde:02}`: Die Zahl wird mit mindestens zwei Stellen ausgegeben, fehlende Stellen werden mit 0 aufgefüllt. Ändern Sie Ihre Ausgabe entsprechend.

```
Die Maschine ist am Tag 1 um 01:15 Uhr fertig.
```

````python
start_stunde = int(input("Startstunde: "))
start_minute = int(input("Startminute: "))
laufzeit = int(input("Laufzeit in Minuten: "))

ende = start_stunde * 60 + start_minute + laufzeit

tag = ende // 1440             # 1 Tag = 24 * 60 min = 1440 min
stunde = ende // 60 % 24
minute = ende % 60

print(f"Die Maschine ist am Tag {tag} um {stunde:02}:{minute:02} Uhr fertig.")
````
