Ein **Struktogramm** stellt den Ablauf eines Programms grafisch dar, unabhängig von der Programmiersprache. Es wird **von oben nach unten** gelesen. Bei einer **Fallauswahl** stehen die Bedingungen über den Spalten. Sie werden von links nach rechts geprüft, ausgeführt wird die erste Spalte, deren Bedingung zutrifft. Die Spalte **sonst** gilt, wenn keine Bedingung zutrifft.

3. **Bedingungen verknüpfen.** Mit `and` und `or` lassen sich zwei Bedingungen zu einer verbinden:

| Verknüpfung | ist wahr, wenn … | Beispiel |
|---|---|---|
| `and` | **beide** Bedingungen wahr sind | `tuer == "zu" and notaus == "nein"` |
| `or` | **mindestens eine** Bedingung wahr ist | `tuer == "offen" or notaus == "ja"` |

Eine Maschine darf nur starten, wenn die Schutztür geschlossen und der Not-Aus nicht gedrückt ist. Setzen Sie das Struktogramm in Python um und testen Sie **alle vier** Kombinationen der Eingaben.

![Struktogramm Maschinenfreigabe](img/sg_freigabe.png)

Schreiben Sie das Programm anschließend so um, dass die Bedingung mit `or` geprüft wird, also nach den Gründen für eine Sperre gefragt wird. Das Verhalten soll gleich bleiben.

````python
tuer = input("Schutztür (zu/offen): ")
notaus = input("Not-Aus gedrückt (ja/nein): ")
if tuer == "zu" and notaus == "nein":
    print("Maschine startet")
else:
    print("Start gesperrt")

# Dieselbe Prüfung mit or: gesperrt, sobald einer der beiden Gründe vorliegt
if tuer == "offen" or notaus == "ja":
    print("Start gesperrt")
else:
    print("Maschine startet")
````

## Make – Erstellen

Ein Sachbearbeiter gibt den Listenpreis eines Artikels, die gekaufte Menge und den Kundenstatus ein. Liegt der Wert des Einkaufs **über 10.000 €**, erhält der Kunde einen Rabattsatz von 15 %, ansonsten 10 %. Ist der Kunde „Stammkunde“, erhält er zusätzlich 4 % Rabatt, als „treuer Kunde“ zusätzlich 2 %. Das Programm berechnet die Höhe des Rabatts in Euro und gibt sie aus.

1. Vervollständigen Sie das Struktogramm auf dem Arbeitsblatt.
2. Setzen Sie das Struktogramm in Python um.

So könnte die Ausgabe aussehen:

```
Listenpreis in €: 100
Menge: 50
Kundenstatus: Stammkunde
Rabatt: 700.0 €
```

### Testen und Kommentieren

| Listenpreis | Menge | Kundenstatus | erwartete Ausgabe |
|---|---|---|---|
| 100 | 50 | Stammkunde | Rabatt: 700.0 € |
| 500 | 20 | treuer Kunde | Rabatt: 1200.0 € |
| 1000 | 11 | Neukunde | Rabatt: 1650.0 € |
| 2000 | 10 | Stammkunde | Rabatt: 3800.0 € |

Kommentieren Sie die beiden Auswahlen in Ihrem Programm.

````python
listenpreis = float(input("Listenpreis in €: "))
menge = int(input("Menge: "))
kundenstatus = input("Kundenstatus: ")
wert = listenpreis * menge

# Grundrabatt nach dem Wert des Einkaufs
if wert > 10000:
    rabatt = 15
else:
    rabatt = 10

# Zusatzrabatt nach dem Kundenstatus (Fallauswahl, sonst-Spalte leer)
if kundenstatus == "Stammkunde":
    rabatt = rabatt + 4
elif kundenstatus == "treuer Kunde":
    rabatt = rabatt + 2

rabattbetrag = wert * rabatt / 100
print(f"Rabatt: {rabattbetrag} €")
````
