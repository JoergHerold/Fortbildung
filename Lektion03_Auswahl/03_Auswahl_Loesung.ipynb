## Predict – Vorhersagen

Das folgende Programm überwacht die Temperatur einer Maschine. Lesen Sie es, **ohne es auszuführen**. Welche Ausgabe erwarten Sie bei den Eingaben **95**, **75**, **70** und **20**?

Notieren Sie Ihre Vorhersagen in der Tabelle auf dem Arbeitsblatt.

````python
# Starterprogramm Lektion 3
temperatur = int(input("Temperatur in °C: "))
if temperatur > 90:
    print("Alarm: Maschine abschalten!")
elif temperatur > 70:
    print("Warnung: Kühlung prüfen")
else:
    print("Temperatur in Ordnung")
print("Messung beendet")
````

## Investigate – Untersuchen

Beantworten Sie die Fragen auf dem Arbeitsblatt. In der Zelle unten steht eine Kopie des Starterprogramms. Dort dürfen Sie alles ändern und ausprobieren, das Original oben bleibt erhalten.

1. Bei der Eingabe 95 sind beide Bedingungen `temperatur > 90` und `temperatur > 70` erfüllt. Warum erscheint trotzdem nur eine Meldung?
2. Vertauschen Sie die beiden Bedingungen, sodass zuerst `temperatur > 70` und danach `temperatur > 90` geprüft wird. Was erscheint jetzt bei 95? Woran liegt das?
3. Machen Sie die Vertauschung rückgängig und ersetzen Sie `elif` durch `if`. Was ändert sich bei der Eingabe 95?
4. Fügen Sie direkt nach der Eingabe die Zeile `print(temperatur > 70)` ein und geben Sie 75 und dann 20 ein. Was ist eine Bedingung?
5. Schreiben Sie statt `if temperatur > 90:` einmal `if temperatur = 90:`. Was meldet Python? Worin unterscheiden sich `=` und `==`?
6. Rücken Sie die letzte Zeile `print("Messung beendet")` ein. Bei welchen Eingaben erscheint sie jetzt? Welche Aufgabe haben der Doppelpunkt und die Einrückung?

````python
# Kopie des Starterprogramms mit den Versuchen aus Investigate
temperatur = int(input("Temperatur in °C: "))
print(temperatur > 70)           # Frage 4: liefert True oder False
if temperatur > 90:              # Frage 5: "=" statt ">" führt zu einem SyntaxError
    print("Alarm: Maschine abschalten!")
elif temperatur > 70:            # Frage 3: mit "if" statt "elif" erscheinen bei 95 Alarm UND Warnung
    print("Warnung: Kühlung prüfen")
else:
    print("Temperatur in Ordnung")
print("Messung beendet")         # Frage 6: eingerückt gehört die Zeile zum else-Zweig
````

## Modify – Verändern

Arbeiten Sie die Schritte der Reihe nach ab. Kopieren Sie das Starterprogramm in die Zelle unten.

1. Im Alarmfall soll zusätzlich ausgegeben werden, um wie viel Grad die Grenze von 90 °C überschritten ist, zum Beispiel „Grenze um 5 °C überschritten“.
2. Ergänzen Sie einen weiteren Fall: **Unter 5 °C** soll die Meldung „Frostgefahr“ erscheinen. Überlegen Sie, an welcher Stelle der neue Zweig stehen muss.
3. Fragen Sie am Ende des Programms: „Wurde die Maschine heute gewartet? (j/n)“. Geben Sie bei „j“ eine Bestätigung aus und sonst den Hinweis, dass eine Wartung fällig ist. Probieren Sie auch die Eingabe „J“ aus.

````python
temperatur = int(input("Temperatur in °C: "))
if temperatur > 90:
    print("Alarm: Maschine abschalten!")
    print(f"Grenze um {temperatur - 90} °C überschritten")   # Schritt 1: zweite Zeile im selben Block
elif temperatur > 70:
    print("Warnung: Kühlung prüfen")
elif temperatur < 5:                 # Schritt 2: vor else, sonst wird der Zweig nie erreicht
    print("Frostgefahr")
else:
    print("Temperatur in Ordnung")
print("Messung beendet")

# Schritt 3: Textvergleich ist exakt, "J" landet im else-Zweig
wartung = input("Wurde die Maschine heute gewartet? (j/n) ")
if wartung == "j":
    print("Danke, Wartung ist eingetragen.")
else:
    print("Hinweis: Die Wartung ist fällig.")
````

### Erweiterung

1. Ein Gewicht von 0 kg oder weniger ist keine sinnvolle Eingabe. Geben Sie in diesem Fall „Ungültige Eingabe“ aus.
2. Für versendbare Pakete wird zusätzlich gefragt, ob Expressversand gewünscht ist (j/n). Express kostet 4,50 € Aufschlag.

```
Gewicht in kg: 3.2
Expressversand? (j/n) j
Versandkosten: 11.5 €
```

````python
gewicht = float(input("Gewicht in kg: "))
if gewicht <= 0:                # ungültige Eingabe
    print("Ungültige Eingabe")
elif gewicht > 31.5:            # zu schwer
    print("nicht versendbar")
else:                           # versendbar: Preis bestimmen, dann Express prüfen
    if gewicht <= 2:
        preis = 5
    elif gewicht <= 5:
        preis = 7
    elif gewicht <= 10:
        preis = 10
    else:
        preis = 16
    express = input("Expressversand? (j/n) ")
    if express == "j":
        preis = preis + 4.5
    print(f"Versandkosten: {preis} €")
````

### Zusatzaufgabe: Paketmaße prüfen

Neben dem Gewicht zählen beim Paketdienst auch die Maße.

- Lesen Sie Länge, Breite und Höhe eines Pakets in cm ein, in beliebiger Reihenfolge.
- Ermitteln Sie die längste Seite und das **Gurtmaß** = längste Seite + 2 · (Summe der beiden anderen Seiten).
- Ist die längste Seite über 200 cm oder das Gurtmaß über 300 cm, ist das Paket zu groß. Ist die längste Seite über 120 cm, gilt es als Sperrgut mit 20 € Zuschlag. Sonst ist es ein Normalpaket.

*Tipp:* Mit `and` verknüpfen Sie zwei Bedingungen, zum Beispiel `if laenge >= breite and laenge >= hoehe:`.

| Länge | Breite | Höhe | erwartete Ausgabe |
|---|---|---|---|
| 60 | 40 | 30 | Normalpaket |
| 30 | 150 | 40 | Sperrgut: 20 € Zuschlag |
| 100 | 80 | 60 | Paket zu groß |
| 10 | 210 | 10 | Paket zu groß |

````python
laenge = float(input("Länge in cm: "))
breite = float(input("Breite in cm: "))
hoehe = float(input("Höhe in cm: "))
if laenge >= breite and laenge >= hoehe:   # Länge ist die längste Seite
    laengste = laenge
elif breite >= hoehe:                      # sonst ist es Breite oder Höhe
    laengste = breite
else:
    laengste = hoehe
gurtmass = laengste + 2 * (laenge + breite + hoehe - laengste)
if laengste > 200:
    print("Paket zu groß")
elif gurtmass > 300:
    print("Paket zu groß")
elif laengste > 120:
    print("Sperrgut: 20 € Zuschlag")
else:
    print("Normalpaket")
````
