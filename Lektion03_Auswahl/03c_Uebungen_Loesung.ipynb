## Aufgabe 1: Kartons packen

In der Versandabteilung werden Bauteile in Kartons zu je **12 Stück** verpackt. Schreiben Sie ein Programm, das die Stückzahl einliest und ausgibt, ob die Teile genau aufgehen. Falls nicht, soll das Programm ausgeben, wie viele Kartons nötig sind und wie viele Teile im letzten Karton liegen.

So könnte die Ausgabe aussehen:

```
Stückzahl: 64
Es werden 6 Kartons benötigt, im letzten liegen 4 Teile.
```

| Stückzahl | erwartete Ausgabe |
|---|---|
| 60 | Die Teile passen genau in 5 Kartons. |
| 64 | Es werden 6 Kartons benötigt, im letzten liegen 4 Teile. |
| 30 | Es werden 3 Kartons benötigt, im letzten liegen 6 Teile. |

*Hinweis:* Denken Sie an `//` und `%` aus Lektion 2.

````python
stueck = int(input("Stückzahl: "))
if stueck % 12 == 0:                 # kein Rest: alle Kartons sind voll
    print(f"Die Teile passen genau in {stueck // 12} Kartons.")
else:                                # ein angebrochener Karton kommt dazu
    kartons = stueck // 12 + 1
    print(f"Es werden {kartons} Kartons benötigt, im letzten liegen {stueck % 12} Teile.")
````

## Aufgabe 2: Notenpunkte umrechnen

In der Oberstufe werden Leistungen in Notenpunkten von 0 bis 15 angegeben. Schreiben Sie ein Programm, das die Punkte einliest und die zugehörige Note ausgibt. Eingaben außerhalb von 0 bis 15 sollen als ungültig gemeldet werden.

| Notenpunkte | 15–13 | 12–10 | 9–7 | 6–4 | 3–1 | 0 |
|---|---|---|---|---|---|---|
| Note | 1 | 2 | 3 | 4 | 5 | 6 |

Testen Sie mindestens mit 15, 13, 12, 4, 1, 0, 16 und -1. Notieren Sie zu jedem Testfall die erwartete Ausgabe als Kommentar unter Ihrem Programm.

````python
punkte = int(input("Notenpunkte: "))
if punkte > 15:            # über 15: ungültig
    print("Ungültige Eingabe")
elif punkte >= 13:         # 13 bis 15
    print("Note 1")
elif punkte >= 10:         # 10 bis 12
    print("Note 2")
elif punkte >= 7:          # 7 bis 9
    print("Note 3")
elif punkte >= 4:          # 4 bis 6
    print("Note 4")
elif punkte >= 1:          # 1 bis 3
    print("Note 5")
elif punkte == 0:          # genau 0
    print("Note 6")
else:                      # negativ: ungültig
    print("Ungültige Eingabe")

# 15 -> Note 1, 13 -> Note 1, 12 -> Note 2, 4 -> Note 4, 1 -> Note 5
# 0 -> Note 6, 16 -> Ungültige Eingabe, -1 -> Ungültige Eingabe
````

## Aufgabe 4: Werkstattrechnung (Struktogramm umsetzen)

Setzen Sie das Struktogramm in Python um. So sehen auch die Struktogrammaufgaben in der Kurzarbeit aus.

![Struktogramm Werkstattrechnung](img/sg_werkstatt.png)

| Stunden | Material in € | erwartete Ausgabe |
|---|---|---|
| 2 | 30 | 150.0 |
| 1 | 20 | 100 |
| 15 | 200 | 1045.0 |

*Zum Nachdenken:* Bei welcher Rechnung greift der Mindestbetrag? Woran erkennen Sie im Struktogramm, dass hier im Code kein `else` nötig ist?

````python
stunden = float(input("Arbeitsstunden: "))
material = float(input("Materialkosten in €: "))
arbeitslohn = stunden * 60
summe = arbeitslohn + material
if summe > 1000:                 # Großauftrag: 5 % Rabatt
    rabatt = summe * 0.05
else:
    rabatt = 0
endbetrag = summe - rabatt
if endbetrag < 100:              # einseitige Auswahl: Mindestbetrag, kein else
    endbetrag = 100
print(endbetrag)
````

## Aufgabe 5: Schaltjahr

Ein Jahr ist ein **Schaltjahr**, wenn es durch 4 teilbar ist, aber nicht durch 100. Jahre, die durch 400 teilbar sind, sind trotzdem Schaltjahre. Schreiben Sie ein Programm, das eine Jahreszahl einliest und ausgibt, ob es sich um ein Schaltjahr handelt.

| Jahr | erwartete Ausgabe |
|---|---|
| 2024 | Schaltjahr |
| 2023 | kein Schaltjahr |
| 1900 | kein Schaltjahr |
| 2000 | Schaltjahr |

*Hinweis:* „durch 4 teilbar“ heißt `jahr % 4 == 0`.

````python
jahr = int(input("Jahr: "))
if (jahr % 4 == 0 and jahr % 100 != 0) or jahr % 400 == 0:
    print("Schaltjahr")
else:
    print("kein Schaltjahr")
````

## Aufgabe 6 (Herausforderung): Quadratische Gleichung

Eine quadratische Gleichung hat die Form $ax^2 + bx + c = 0$. Ob und wie viele Lösungen es gibt, entscheidet die **Diskriminante** $d = b^2 - 4ac$:

- $d > 0$: zwei Lösungen
- $d = 0$: eine Lösung
- $d < 0$: keine Lösung

Die Lösungen berechnet man mit $$x_{1,2} = \frac{-b \pm \sqrt{d}}{2a}$$

Schreiben Sie ein Programm, das a, b und c einliest und ausgibt, wie viele Lösungen es gibt und welche.

*Hinweis:* Die Wurzel einer Zahl berechnen Sie mit `d ** 0.5`.

| a | b | c | erwartete Lösungen |
|---|---|---|---|
| 1 | -3 | 2 | x1 = 2.0, x2 = 1.0 |
| 1 | 2 | 1 | x = -1.0 |
| 1 | 0 | 1 | keine Lösung |

*Erweiterung:* Was passiert bei a = 0? Fangen Sie diesen Fall ab.

````python
a = float(input("a: "))
b = float(input("b: "))
c = float(input("c: "))
if a == 0:  # Erweiterung: a = 0 abfangen
    print("a darf nicht 0 sein.")
else:
    d = b ** 2 - 4 * a * c  # Diskriminante, wird unten wiederverwendet
    if d > 0:
        x1 = (-b + d ** 0.5) / (2 * a)
        x2 = (-b - d ** 0.5) / (2 * a)
        print(f"Zwei Lösungen: x1 = {x1}, x2 = {x2}")
    elif d == 0:
        x = -b / (2 * a)
        print(f"Eine Lösung: x = {x}")
    else:
        print("Keine Lösung")
````
